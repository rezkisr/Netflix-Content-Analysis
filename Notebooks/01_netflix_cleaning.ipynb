```python
import pandas as pd
import numpy as np
import sweetviz as sv
```

```python
df_ori = pd.read_csv('../data/netflix_titles.csv')
```

```python
df = df_ori.copy()
```

```python
report = sv.analyze(df)
report.show_html('../report_netflix.html')
```

```python
for col in df.select_dtypes(include=['object', 'string']):
    df[col] = df[col].str.strip()
```

```python
df['country'] = df['country'].fillna(df['country'].mode()[0])
df['director'] = df['director'].fillna('No Data')
df['cast'] = df['cast'].fillna('No Data')
```

```python
df= df.dropna(subset={'date_added', 'duration', 'rating'})
```

### Standardize Audience Ratings

The original Netflix maturity ratings were grouped into broader audience categories to simplify analysis and visualization. The grouping is based on Netflix's official maturity rating definitions for the United States.

**Reference:**
https://help.netflix.com/en/node/2064/us

```python
df['rating'] = df['rating'].replace({
    'TV-Y7-FV' : 'TV-Y7'
})

df['rating'] = df['rating'].replace({
    'PG-13' : 'Teens - Age above 12',
    'TV-MA' : 'Adults',
    'PG' : 'Kids - with parental guidance',
    'TV-14' : 'Age above 14',
    'TV-PG' : 'Kids - with parental guidance',
    'TV-Y' : 'Kids',
    'TV-Y7' : 'Kids - Age above 7',
    'R' : 'Adults',
    'TV-G' : 'Kids',
    'G' : 'Kids',
    'NC-17' : 'Adults',
    'NR' : 'Not Rated',
    'UR' : 'Unrated'
})
```

```python
df['date_added'] = pd.to_datetime(df['date_added'])
```

```python
df = df.reset_index(drop=True)
```

# DONE

```python
df.to_csv('../data/Netflix_cleaned.csv', index= False)
```
